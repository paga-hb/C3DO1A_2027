[Docker](https://www.docker.com) is a software platform that allows you to build, test, and deploy applications quickly. Docker packages software into standardized units called **images** that have everything the software needs to run including libraries, system tools, code, and runtime. A Docker **image** acts as a template for Docker **containers**, i.e. a Docker **container** is an instance of a Docker **image**. When we run a containerized application, it is run as a Docker **container**, based on a Docker **image** template.

## Run a Docker Image

We issue the command `docker run` to run an **image**, which creates a **container** using the **image** as a template.

> **Note**: A **container** is an instance of an **image** (a **container** is NOT the same thing as an **image**). We can have multiple **containers** running at the same time, even though they might be instantiated from the same **image**.

The **docker run** command takes a number of aguments. To list the available arguments for any Docker command, we can use the `--help` flag.

Run the command below to print the help for the `run` command.

````python
!docker run --help
````

We can also list all available commands with the command below.

````python
!docker --help
````

Now, let's run our first **container**, based on the `nginx` **image**.

The command `docker run -d --rm --name nginx -p 8080:80 nginx` consists of:

- `docker run` which creates and runs a new **container**
- `-d` which runs the **container** in `detached` mode
  - A **container** running in `detached` mode, is run in a **backgroud** process (we won't "block" our terminal, or see any output in our terminal).
  - If `-d` is omitted, the **container** will run in a **foreground** process (which dedicates our terminal to that process, where we see the output).
- `--rm` stands for `remove`, and will automatically delete the **container** when we stop it from running.
  - If we omit the `--rm` flag, the **container** will NOT be deleted when we stop it from running.
- `--name <ContainerName>` gives our **container** the name `<ContainerName>` which makes it a lot easier to refer to the **container**.
  - `--name nginx` will give our **container** the name `nginx`. 
  - If we omit the `--name` flag, the **container** will be given a randomly generated name.
- `-p` stands for `publish`, which "publishes" ports, i.e. is how we can map a port on our **host** machine to a port in the **container**.
  - `-p <HostPort>:<ContainerPort>` will port forward port `<HostPort>` on our **host** machine to port `<ContainerPort>` in the **container**.
  - `-p 8080:80` will port forward port `8080` on our **host** (local computer) to port `80` in the **container**.
  - `nginx`, running in the **container**, exposes its web server on port `80` as default.
  - This means we can access the `nginx` web server running in the **container** via `http://localhost:8080` on our **host** (local computer).
- `nginx` is the final agument to the `docker run` command, which is the name of the **image**.
  - Notice that we are creating a **container** named `nginx` (`--name nginx`) from an **image** named `nginx` (the final argument).
  - The final argument `nginx` is equivalent to `nginx:latest`, i.e. with the version tag `:latest`.

When we use the `docker run` command:

- Docker will first check if the **image** with the chosen **version** is available on our local computer.
  - In this case `nginx:latest` is already available on the local computer, since we downloaded it with `docker pull nginx`.
- If the **image** with the chosen **version** isn't avaiable on our local computer, Docker will first download it, and then run the **container**.

Run the command below to create a new `nginx` **container** from the `nginx` **image**.

- Notice the returned **hash** value when we execute the `docker run` command, which is the unique identifier for the **container**.

````python
!docker run -d --rm --name nginx -p 8080:80 nginx
````

Now, let's access the web server on http://localhost:8080.

You should see a web page with the text `Welcome to nginx!`.

## List Running Containers

Let's see what **containers** are currently running on our local computer.

This is done with the command `docker ps` (where `ps` stands for `process`).

- The `CONTAINER ID` column displays the unique identifier (hash) for the **container**.
- The `IMAGE` column contains the name of the **image** the **container** is basen on.
- The `COMMAND` column contains the command that was run inside the **container** when it started (more on this later).
- The `CREATED` column shows when the **container** was created (i.e. run for the first time).
- The `STATUS` column displays the **container's** status (`up` means the container is running, with information about how long it's been running).
- The `PORTS` column shows the port mapping (port forwarding) that we configured when running the **container**.
  - `8080` is the port for the **host** (local computer), where `0.0.0.0` if the interface used.
    - An interface of `0.0.0.0` means the **host** is listening on ALL interfaces (e.g. on ALL network cards, if we have more than one).
    - An interface of `127.0.0.1` would mean the **host** is listening on the local loopback interface (i.e. localhost).
    - An interface of `localhost` is the same as `127.0.0.1`, i.e. that the **host** is listening on the local loopback interface.
  - `80` is the port for the **container**.
  - `/tcp` means the port is a TCP port (as opposed to a UDP port).
  - This is a great way of finding out what **host** ports have been mapped (forwarded) to what ports inside the **container**.
- The `NAMES` column shows the name we assigned to the **container** when we started (ran) it.
  - If we hadn't explicitly named our container with `--name nginx` when we started (ran) it, it would have a randomly generated name.

Also check:

- If you have Docker Desktop installed, under `Containers`, you will also see the `nginx` **container** running (a green sold circle).
- You can also start and stop containers from the VSCode Docker extension or via Docker Desktop.

## View the Logs for a Running Container

The approach we use to view the logs of a running **container**, depends on how we started the **container**:

- If we are running a **container** in **attached** mode, i.e. we didn't start the **container** in **detached** mode with the `-d` flag, then all output, including logs, from applications running in the **container** will be printed to our terminal.
- If we are running a **container** in **detached** mode, i.e. we started the **container** with the `-d` flag, then all output, including logs, won't be visible in our terminal.

If we need to access a running **container**'s logs in detached mode, we can:

- Open a terminal on our **host** (local) computer (Menu: `Terminal -> New Terminal`, Windows/Linux: `Ctrl + J`, Mac: `Cmd + J`).
- Running the command `docker logs <ContainerName>`, where:
  - `docker logs` is the command for viewing logs in a running **container**.
  - `<ContainerName>` is the name of the running **container**.
    - In our case, the **container** is named `nginx`.

Now, open a terminal and run the command `docker logs nginx`.

- You will see logs entries for the Nginx Web Server.
- If you had your own application (microservice) running in a container, your would see your own application's output (logs)
  - E.g. the output from every `Console.WriteLine()` in a .NET (C#) application.

## Establish an Interactive Terminal Session with a Running Container

If we need to access a running **container**, we can establish an interactive terminal session by:

- Opening a terminal on our **host** (local) computer (Menu: `Terminal -> New Terminal`, Windows/Linux: `Ctrl + J`, Mac: `Cmd + J`).
- Running the command `docker exec -it <ContainerName> <Command>`, where:
  - `docker exec` executes a command in a running **container**.
  - `-it` consists of the two flags `-i` (for interactive) and `-t` (for tty), which means we want an interactive session.
  - `<ContainerName>` is the name of the running **container**.
    - In our case, our **container** is named `nginx`.
  - `<Command>` is the command we want to execute in the running **container**.
    - In our case we want the command `bash`, which starts a new bash shell inside the **container**.
    - Note that some **containers** don't have a `bash` shell, but only the `sh` shell (the Borne Shell), so we would use `sh` instead of `bash`.

Now, open a terminal and run the command:

```bash
docker exec -it nginx bash
```

- You will see the terminal's prompt change to something like `root@8ac280d9b96d:/#` (this is the `bash` terminal inside the **container**).
- Issue the command `pwd` to see what folder you are in (which will be `/`, i.e. the root folder on a Linux-based **container**).
  - Note that most **containers** (i.e. **images**) are Linux-based, so we need to know how to use Linux shell commands when using **containers**.
  - Linux-based file systems have a root folder called `/`, which is the equivalent of `C:\` on a Windows machine.
- Now, let's list the contents of Nginx's default web site, which is located in the folder `/usr/share/nginx/html`.
- Issue the command `ls -al /usr/share/nginx/html` which will list the contents in the folder `/usr/share/nginx/html`.
  - On a Posix-based operating system (e.g. Linux or Mac), the `ls` command is equivalent to the `dir` command on Windows.
  - `-al` is a combination of the two flags `-a` and `-l`, which will show additional file properties, including hidden files.
  - Notice the file `index.html`, which is Nginx's default web page.
- Issue the command `cat /usr/share/nginx/html/index.html` which will print out the contents of the HTML file.
  - We should see the same information as when we visited `http://localhost:8080` before.

To exit the interactive terminal session, issue the command `exit`.

- Notice the terminal now show's a prompt for the **host** (local) computer again.

## Connecting a Container to a Network

Containers can be connected to one or more **networks**, which is an important concept if we want **containers to communicate with each other**.

The most common type of network is a **bridge**, where:
- All containers (services) on the same **bridge** network can communicate via IP Addresses.
- All containers (services) on the same **bridge** network can communicate via DNS names.
  - A DNS name is automatically created for a container's name.
    - E.g. in `docker run --name <CONTAINER_NAME>`, a DNS name `<CONTAINER_NAME>` would be created for the container.
  - An additional DNS name can be created with the `--hostname` flag.
    - E.g. in `docker run --hostname <HOST_NAME>`, a DNS name `<HOST_NAME>` would be created for the container.
- If a container with DNS name `one` is on the same **bridge** network as a container with DNS name `two`:
  - Container `one` can communicate with conatiner `two` via e.g. `http://two` (or by replacing the DNS name `two` with the IP Address).
  - Container `two` can communicate with conatiner `one` via e.g. `http://one` (or by replacing the DNS name `one` with the IP Address).
  - Container `one` and `two` can also communicate with the host and reach the internet via the host.

**Note!**

- Docker's **default network** is a **bridge** network called **bridge**.
- If you don't specify a network when creating a container, it will be placed on this **default network**.
- All containers on the **default network** can communicate via IP Addresses, but **can not communicate via DNS names**.

When working with networks, we can:
- Create a network with the command `docker network create <NETWORK_NAME>`, where `<NETWORK_NAME>` is the name of the network.
  - This will create a network of type **bridge** be default.
- List all Docker networks on the local computer with the command `docker network ls`.
- Delete a network with the command `docker network rm <NETWORK_NAME>`, where `<NETWORK_NAME>` is the name of the network.
  - Note that you can't delete a network as long as there is at least one running container using it.
- Connect a container to a network using the `--network` flag in a `docker run` or `docker create` command.
  - For example in the command `docker run -d --rm --name nginx --network flixtube -p 8080:80 nginx`
    - `--network flixtube` means we are connecting the container to a network called `flixtube`.

## Mapping Volumes to Containers

**Volumes** are used to persist **container** state (data) outside the **container** so that when a container is removed and created again, the state (data) is still available (otherwise, the state (data) would be lost when the **container** is removed).

When starting a **container**, we can choose to share a folder on the **host** computer with the **container**, using either a:

- **Bind Mount**
  - We bind a folder path on the **host** to a folder path in the **container**.
  - `<HOST_FOLDER_PATH>:<CONTAINER_FOLDER_PATH>`
  - Any files and subfolders added/deleted/modified in the **host** folder is reflected in the **container** folder.
  - Any files and subfolders added/deleted/modified in the **container** folder is reflected in the **host** folder.
  - So, a bind mount just means that we are sharing a folder between the **host** and **container**.
- **Named Volume**
  - We associate the **name** of a volume on the **host** with a folder path in the **container**.
    - `<VOLUME_NAME>:<CONTAINER_FOLDER_PATH>`
  - Docker will create a folder somewhere on the **host** and associate the folder path with the volume **name**.
  - Any files and subfolders added/deleted/modified in the **host** folder is reflected in the **container** folder.
  - Any files and subfolders added/deleted/modified in the **container** folder is reflected in the **host** folder.
  - So, a named volume is like a bind mount, but the folder path on the **host** is managed by Docker.

Volumes are mapped when creating (`docker create`) or running (`docker run`) a **container**, via the `-v` flag (where `-v` stands for `--volume`).

- For a **bind mount**: `-v <HOST_FOLDER_PATH>:<CONTAINER_FOLDER_PATH>`
- For a **named volume**: `-v <VOLUME_NAME>:<CONTAINER_FOLDER_PATH>`

## Upload an Image to Docker Hub

Let's upload an **image** to our new repository on Docker Hub.

We'll use the `nginx` **image** we downloaded before as an example.

Start by listing **images** on the local computer again.

- The `nginx` **image** is listed as:
  - `nginx:latest` under `IMAGE`, which is really `docker.io/registry/nginx:latest` but abbreviated to `nginx` since it's a Docker Hub official **image**. 

````python
!docker images
````

### Tag the Image

To upload the `nginx` **image** to our new repository, we need to:

- Tag the local **image** with our own repository's `Domain`, `Namespace` and `Repository Name`.
- Push the newly tagged **image** to our repository.

The command `docker tag <ExistingImage>:<ExistingImageTag> <NewImage>:<NewImageTag>`:

- Creates a copy of the local **image** `<ExistingImage>:<ExistingImageTag>`, where:
  - `<ExistingImage>` is what's shown under the `IMAGE` column above, before the colon `:`, i.e. `nginx`.
  - `<ExistingImageTag>` is what's shown under the `IMAGE` column above, after the colon `:`, i.e. `latest`.
  - `<NewImage>` is made up of our repository's `<Domain>`, `<Namespace>` and `<RepositoryName>`, where:
    - `<Domain>` is `docker.io` for Docker Hub, but can be omitted when we want to upload an image to Docker Hub.
    - `<Namespace>` is our username on Docker Hub.
    - `<RepositoryName>` is our repository's name on Docker Hub, i.e. `nginx` in this case.
  - `<NewImageTag>` is the version we want to give our new **image**, e.g. `1.0`.

The command's structure is thus, `docker tag <ExistingImage>:<ExistingImageTag> <Domain>/<Namespace>/<RepositoryName>:<NewImageTag>`.

To tag your **image**, use the command below, but replace `<Namespace>` with your Docker Hub username.

`docker tag nginx:latest docker.io/<Namespace>/nginx:1.0`

Or, since the default domain is `docker.io`, we can abbreviate the command to:

`docker tag nginx:latest <Namespace>/nginx:1.0`

When you have issued the command above, list all **images** on your local computer again.

- Notice that you now have two **images**, both with the same `ID`, but with:
  - `nginx:latest` under `IMAGE` for the original **image**.
  - `<Namespace>/nginx:1.0` under `IMAGE` for the new **image** copy, where `<Namespace>` is your Docker Hub username.  

````python
!docker images
````

### Login to a Repository (Docker Hub)

Before you can upload your new **image**, you need to login to Docker Hub using the [docker login](https://docs.docker.com/reference/cli/docker/login) command.

- Open a terminal and run the command `docker login -u <Namespace> -p <AccessToken>`, where:
  - `<Namespace>` is your Docker Hub username.
  - `<AccessToken>` is your Docker Hub Access Token (that your previously saved).

Or, use the `--password-stdin` flag instead of the `-p` flag, which will read the `<AccessToken>` from Standard In.

- Open a terminal and run the command `echo <AccessToken> | docker login -u <Namespace> --password-stdin`, where:
  - `<Namespace>` is your Docker Hub username.
  - `<AccessToken>` is your Docker Hub Access Token (that your previously saved).
  - Note that you need to use PowerShell (not CMD) if on Windows.

Or, store store your access token in a text file, then:
- Open a terminal and run the command `cat <PathToTextFile> | docker login -u <Namespace> --password-stdin`, where:
  - `<Namespace>` is your Docker Hub username.
  - `<PathToTextFile>` is the path to your textfile, containing your access token, in your file system.

> **Note**: To logout from Docker Hub, use the command `docker logout` (but don't do this now).

### Upload the Image to the Repository (Docker Hub) 

Now, upload the new **image** to your Docker Hub repository.

This is done using the command `docker push <NewImage>:<NewImageTag>`, where:

- `<NewImage>` is what's listed under `IMAGE`, before the colon `:`, for our new **image**, i.e. `<Namespace>/nginx` where `<Namespace>` is your Docker Hub username.
- `<NewImageTag>` is what's listed under `IMAGE`, after the colon `:`, for our new **image**, i.e. `1.0`.

Issue the command below, replacing `<Namespace>` with your Docker Hub username.

- `docker push <Namespace>/nginx:1.0`

Then visit your repository at `https://hub.docker.com/repository/docker/<Namespace>/nginx`, where `<Namespace>` is your Docker Hub username.

- Under `tags`, notice the newly uploaded `nginx` **image** with tag (version) `1.0`.

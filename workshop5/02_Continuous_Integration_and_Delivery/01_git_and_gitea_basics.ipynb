Let's see how Git is configured using Git commands.
- `git config --list --system` shows us system settings (for ALL users on the computer).
- `git config --list --global` shows us global settings (for the logged in user, i.e. `student` if you are using one of the university's VMs).
- `git config --list --local` shows us local settings (for the current local Git repository, i.e. the folder (`test` in this case) containing the `.git` subfolder).
- `git config --list --show-origin` shows us all settings (system, global, and local - in that order).
  - We see that (on one of the university's VMs):
    - System settings are stored in the file `/etc/gitconfig`
    - Global settings are stored in the file `/home/student/.gitconfig`
    - Local settings are stored in the file `.git/config` (under the folder `test`)

**Note**
- According to Git's precedence rules:
  - A local setting overrides a global and system setting.
  - A global setting overrides a system setting.
- To edit Git configuration files from a terminal, we could use the commands:
  - `git config --system -e`
  - `git config --global -e`
  - `git config --local -e`

````python
%%bash

cd test

echo -e "\n--- System git config ---\n"
git --no-pager config --list --system

echo -e "\n--- Gloabl git config ---\n"
git --no-pager config --list --global

echo -e "\n--- Local git config ---\n"
git --no-pager config --list --local

echo -e "\n--- All git config ---\n"
git --no-pager config --list --show-origin
````

Let's configure local settings for our local Git repository in the folder `test`.
- Since we are issuing the `git config` commands from within the `test` folder, they will be `--local` by default (to make them global, we would use the `--gloabl` flag).
- These local settings will override any global and system settings, and will be used for any Git command as long as we are issuing the Git commands from within the `test` folder (otherwise Git will use global settings by default).

Settings:
- `git config user.name "John Doe"`
  - This sets the Git user's full name.
- `git config user.email "john.doe@gmail.com"`
  - This sets the Git user's email address.
- `git config credential.username john.doe`
  - This sets the Git user's username (i.e. the Gitea user's username).
- `git config credential.helper "store --file .git/credentials"`
  - This setting will store any Git credentials in the file `.git/credentials` (under the `test` folder).
  - There are many ways to store Git credentials, but we'll use this so we can view the actual Git credentials stored.
- `git config --unset credential.useHttpPath`
  - This sets `useHttpPath=false`.
  - This setting tells Git how to store credentials for repositories.
    - `useHttpPath=false` (default)
      - Git stores username and password `user:pass` per HOST, e.g.:
        - http://user:pass@example.com
    - `useHttpPath=true`
      - Git stores username and password `user:pass` per FULL URL (per repo), e.g.: 
        - http://user1:pass1@example.com/repo1.git
        - http://user2:pass2@example.com/repo2.git
- `echo 'http://john.doe:john.doe@192.168.122.99:3000' > .git/credentials`
  - Here we are manually storing Git credentials in our credential store (i.e. the file `.git/credentials`), which means we are:
    - Comminicating with our Gitea server `http://192.168.122.99` on port `3000`.
    - Using `john.doe` as the Gitea user's username (this is the first `john.doe` before the colon `:`).
    - Using `john.doe` as the Gitea user's password (this is the second `john.doe` after the colon `:`).
  - Note that we are using single quotes `echo ''` and not doubble quotes `echo ""`
    - This is so that the command still works if the password contains special characters, e.g. `!`.
- `chmod 600 .git/credentials`
  - Here we are setting the required permissons for the file `.git/credentials`.

**Note!**
- If we didn't manually add Git credentials to `.git/credentials`, Git would ask for our password `john.doe` the first time we tried to `push` something to Gitea, and then store the password in `.git/credentials`.
- If we want to store our credentials (and user details) globally (for the user) instead of locally (just for the local Git repository), we would use:
  - This would store the credentials in the file `~/.git-credentials` on Linux.

  ```bash
  git config --global user.name "John Doe"
  git config --global user.email "john.doe@gmail.com"
  git config --global credential.username john.doe
  git config --global credential.helper store
  git config --global --unset credential.useHttpPath
  ```

````python
%%bash

cd test

git config user.name "John Doe"
git config user.email "john.doe@gmail.com"
git config credential.username john.doe
git config credential.helper "store --file .git/credentials"
git config --unset credential.useHttpPath
echo 'http://john.doe:john.doe@192.168.122.99:3000' > .git/credentials
chmod 600 .git/credentials
````

### Working in a New Feature Branch and Using Pull Requests
- We can switch between branches in our local repository with the command `git switch <branch>`, where `<branch>` is the name of the branch.
  - For example, the command `git switch main` switches to the `main` branch.
- To get the latest changes from the remote repository's branch, we can use the command `git pull`.
- We can create a new branch called `feature` in our local repository, and then switch to it, with the command `git switch -c feature`
  - The `-c` flag tells Git to create the local branch `feature` before switching to it.
- We can then work in the `feature` branch as usual.
  - Below, we are creating a new empty file `hello.txt`, staging it with `git add .`, and commiting it with `git commit -m "add hello.txt"`.
- To push the local branch `feature` to the remote repository, we need to use the command `git push -u origin feature`.
  - Here we are using `git push` that we are familiar with.
  - `-u origin` sets the *upsteam* (remote) repository `-u` to `origin` in our local `feature` branch.
    - `origin` is a Git alias for `http://192.168.122.99:3000/MyOrg/testrepo.git` (which we can see with the command `git remote -v`).
    - The first time we use a `git push` in a new branch, we need to tell Git what the upsteam (remote) repository is.
    - After that, we can skip `-u origin` when issuing `git push` commands, since Git remembers the setting.
  - `feature` tells Git that we want to push to the remote branch `feature` on the Gitea server.
    - Since the branch `feature` doesn't exist on the Gitea server yet, it will be created.

Now, let's run the cell below that does all of this.

````python
%%bash

cd test
git switch main
git pull
git switch -c feature
touch hello.txt
git add .
git commit -m "add hello.txt"
git push -u origin feature
````

### Working in a Local Feature Branch and Merging into Main without a Pull Request

### Login to Gitea via the CLI 

- Remember that we created a token for John on Gitea (near the top of this notebook).
  - `Settings` $\rightarrow$ `Applications` $\rightarrow$ `Generate Token`
  - The token looks something like this: `19856996a7e498aebd8c51760c0ff889b940503c`
- We can add a login to Gitea using the command `tea login add -n <username> -u <URL> -t <token> -i`
  - `tea login add` is the command we use to add a new login to Tea.
  - `-n <username>` specifies the Gitea user's username, where `<username>` is the username.
  - `-u <URL>` specifies the Gitea server, where `<URL>` consists of `<scheme>://<IP>:<port>`, e.g.`http://192.168.122.99:3000`
  - `-t <token>` specifies the Gitea user's access token, where `<token>` is the access token.

**Note**
- Make sure you replace the token in the cell below with YOUR token, before executing the cell.
- We can also remove a login we previously added with `tea login rm <username>`

````python
%%bash

tea login add -n john.doe -u http://192.168.122.99:3000 -t 19856996a7e498aebd8c51760c0ff889b940503c -i
# tea login rm john.doe
````

## Using Gitea Container Registry

In a previous workshop, we saw how we could upload and download Docker Images from Docker Hub.

Gitea has a Container Registry, where we can also upload and download Docker Images.

Let's see how we can Upload/Download images to/from our `testrepo` repository under our `MyOrg` organization.
- Images can be uploaded/downloaded to/from an organization.
- Images can be uploaded/downloaded to/from a repository.

#### Download Nginx from Docker Hub
- Let's download the latest version of the `Nginx` image from Docker Hub.
  - We'll also list the local images, so we see that the images was downloaded.

````python
!docker pull nginx:latest
!docker images
````

#### Tag the Image
We need to tag the Nginx image with our Gitea details.

- Syntax:
  - If we want to upload the image to an organization, we would use:
  
    ```bash
    docker tag <ExistingImage>:<ExistingImageTag> <Domain>/<Namespace>:<NewImageTag>
    ```
  
  - If we want to upload the image to a repository, we would use:
  
    ```bash
    docker tag <ExistingImage>:<ExistingImageTag> <Domain>/<Namespace>/<RepositoryName>:<NewImageTag>
    ```

- Let's tag the Nginx image with a version of `1.0`.
  - If we want to upload the image to `MyOrg`, we would use:

    ```bash
    docker tag nginx:latest 192.168.122.99:3000/MyOrg/nginx:1.0
    ```
  
  - If we want to upload the image to `MyOrg/testrepo`, we would use:

    ```bash
    docker tag nginx:latest 192.168.122.99:3000/MyOrg/testrepo/nginx:1.0
    ```

Run the cell below to tag the Nginx image for uploading to both the organization `MyOrg` and the repository `MyOrg/testrepo`.
- We also list local images, so we see the original image and the two tagged images.

````python
!docker tag nginx:latest 192.168.122.99:3000/MyOrg/nginx:1.0
!docker tag nginx:latest 192.168.122.99:3000/MyOrg/testrepo/nginx:1.0
!docker images
````

#### Login to Gitea

Next, we need to log Docker into Gitea.

- Syntax:

  ```bash
  docker login -u <User> -p <Password> <URL>
  ```

- Let's login as user `john.doe` with password `john.doe` to our Gitea server `http://172.31.43.21:3000`.

  ```bash
  docker login -u john.doe -p john.doe http://172.31.43.21:3000
  ```

````python
!docker login -u john.doe -p john.doe http://172.31.43.21:3000
````

#### Upload the Images to Gitea Container Registry

Now we can push our tagged images to our Gitea Container Registry.

- Gitea actually places both images under the organization's `Packages` tab in Gitea's Web UI.
  - The image only tagged with the organization `MyOrg`, will show up as `nginx` under the organization's `Packages` tab.

    ```bash
    docker push 192.168.122.99:3000/MyOrg/nginx:1.0
    ```

  - The image tagged with both the organization and the repository `MyOrg/testrepo`, will show up as `MyOrg/nginx` under the organization's `Packages` tab.
    - Note that the image still shows up under the organization's `Packages` tab, and not under the repository's `Packages` tab.

    ```bash
    docker push 192.168.122.99:3000/MyOrg/testrepo/nginx:1.0
    ```
After running the cell below:
- Visit http://192.168.122.99:3000/MyOrg/-/packages
  - Notice that both images have been uploaded under the organization's `Packges` tab.
- Visit http://192.168.122.99:3000/MyOrg/testrepo/packages
  - Notice that no images show up under the repository's `Packges` tab.

````python
!docker push 192.168.122.99:3000/MyOrg/nginx:1.0
!docker push 192.168.122.99:3000/MyOrg/testrepo/nginx:1.0
````

#### Delete the Local Images
- Let's delete the local images.
  - We also list the local images, so we see they have been deleted.

````python
!docker rmi 192.168.122.99:3000/MyOrg/nginx:1.0
!docker rmi 192.168.122.99:3000/MyOrg/testrepo/nginx:1.0
!docker rmi nginx:latest
!docker images
````

## Running Shell Commands from Jupyter Notebook Cells

We can use *cell magics* to run shell commands in Jupyter Notebook cells.
- A Jupyter Notebook cell that starts with a `!`, is used to execute a single command in your operating system's default shell, e.g.
  ```bash
  !ls
  !pwd
  ```
- A Jupyter Notebook cell that starts with `%%bash` is used to execute multiple commands in your operating system's bash shell, e.g.
  ```bash
  %%bash
  ls
  pwd
  ```
- A Jupyter Notebook cell that starts with `%%zsh` is used to execute multiple commands in your operating system's zsh shell, e.g.
  ```bash
  %%zsh
  ls
  pwd
  ```
- A Jupyter Notebook cell that starts with `%%cmd` is used to execute multiple commands in your operating system's CMD shell, e.g.
  ```bash
  %%cmd
  dir
  pwd
  ```
- A Jupyter Notebook cell that starts with `%%pwsh` is used to execute multiple commands in your operating system's PowerShell, e.g.
  ```bash
  %%pwsh
  dir
  pwd
  ```

- Of couse, the type of shell must exist on your operating system.
  - It's safe to use `!` on any operating system.
  - It's safe to use `%%bash` on Linux and Mac.
  - It's safe to use `%%cmd` on Windows.
- If you see a cell containing `%%bash`, in any notebook in this course, and you are on Windows, change it to `%%cmd`

## Docker Desktop (Windows and Mac)

**Note! If you are on Linux (or a university virtual machine), skip this step.**

Let's verify Docker Desktop is set up correctly.

- Search for `Docker Desktop` among the installed applications on your computer and start it.
  - If you are on Linux, and have chosen to install `Docker Engine` instead of `Docker Desktop`, ensure the  `Docker Engine` daemon is running.
- In the bottom left of `Docker Desktop's` user interface, wait until the status text `Engine starting` changes to `Engine running`.
  - If you are on Windows, and prompted to set up Windows Subsystem for Linux (WSL), follow the instructions to do so.
- Click the `cog wheel` icon in the top right of `Docker Desktop's` user interface.
  - Under the `General` tab, make sure the checkbox is checked for `Start Docker Desktop when you sign in to your computer`.
  - Under the `Kubernetes` tab, make sure `Enable Kubernetes` is **disabled** (not enabled).
    - We will **not** be using Docker Desktop's Kubernetes cluster in this course (we will be using Minikube instead).
  - Click the `Apply & restart` button in the lower right of `Docker Desktop's` user interface.
  - When `Docker Desktop` has restarted, wait until the status text `Engine starting` changes to `Engine running` again.
  - To the right of this text, in the bottom of `Docker Desktop's` user interface, you will see the text `Kubernetes running` if you enabled Docker Desktop's Kubernetes cluster (if so, make sure you **disable** Docker Desktop's Kubernetes cliuster via the settings above).

With the setting above, `Docker Desktop` should start when your computer boots, and run automatically run `Docker Engine`.

- **Note!**: If you didn't check the checkbox for `Start Docker Desktop when you sign in to your computer`, you need to start `Docker Desktop` manually after your computer has booted, before you can use `Docker` or `Docker Compose`.

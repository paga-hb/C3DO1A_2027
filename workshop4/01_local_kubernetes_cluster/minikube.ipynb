## Start a Minikube cluster

The `minikube start` command will start a Minikube cluster (and also create it if doesn't already exist).
- The `--nodes` option specifies the number of Minikube nodes that will be created (one node will be a control plane node).
- The `--cpus` option specifies the number of vCPUs to assign to each Minikube node.
- The `--memory` option specifies the (maximum) amount of RAM to assign to each Minikube node, where `m` stands for `MB`.
- The `--disk-size` option specifies the (maximum) amount of disk space to use for each Minikube node, where `m` stands for `MB`.
- The `--profile` option creates a cluster with a specific name (profile).
  - If no `--profile` option is specified, the cluster will be called `minikube` by default.
  - Multiple clusters with different names (profiles) can be created.

**Note**
- Calling `minikube start` without any options will create a **single-node** Minikube Kubernetes cluster called `minikube`, with:
  - `2` vCPUs
  - `3.9GB` RAM
  - `20GB` disk space
- If you have a limited amount of resources (CPU, RAM, Disk) on your computer, starting a single-node Minikube Kuberentes cluster is the best option.

Let's create and start a 3-node Minikube Kubernetes cluster (called `minikube` by default).

## Remote into a Minikube node

- We can open a Secure SHell (SSH) to a Minikube node with the command `minikube ssh --node <node>`.
  - `<node>` is the name of he Minikube node.  
- For example:
  - `minikube ssh --node minikube`
  - `minikube ssh --node minikube-m02`
  - `minikube ssh --node minikube-m03`

- This can be useful for debugging purposes since, within a Minikube node, we:
  - Have access to the internal (private) Kubernetes network.
  - Can see what Docker containers have been pulled into a Node (e.g. for a Pod running on a worker node).

### Run the commands below in a separate terminal (the Notebook cell won't terminate otherwise).

```bash
# Remote into the master node
minikube ssh --node minikube

# List kubernets system components (processes) running on the master node
ps -e | grep -i -E 'etcd|kube|containerd$' | sort

# containerd      : this is the container runtime engine that manages containers on a node.
# kube-scheduler  : this is kube-scheduler that schedules new resources on specific nodes.
# etcd            : this is the etcd key-value database that stores a Kubernetes cluster's state ("single source of truth").
# kube-apiserver  : this is the kube-apiserver that communicates with the etcd database and exposes a REST API.
# kube-controller : this is the kube-controller-manager that controls various resources' controller managers.
# kubelet         : this is the kubelet that communicates with the kube-apiserver and manages pods on a node.
# kube-proxy      : this is the kube-proxy that maintains a network routing table and maps Service IPs Pod IPs.

# Exit from the master node
exit

# Remote into a worker node
minikube ssh --node minikube-m02

# List kubernets system components (processes) running on the worker node
ps -e | grep -i -E 'etcd|kube|containerd$' | sort

# containerd      : this is the container runtime engine that manages containers on a node.
# kubelet         : this is the kubelet that communicates with the kube-apiserver and manages pods on a node.
# kube-proxy      : this is the kube-proxy that maintains a network routing table and maps Service IPs Pod IPs.

# Exit from the worker node
exit
```

## Enable Minikube addons

- Addons are additional resources that can be added to a specific Minikube cluster.
  - Usually you would have to install these yourself in a normal Kubernetes cluster, but Minikube Addons makes this easy.

- Addons can be enabled with the command `minikube addons enable <addon>`, where `<addon>` is the name of the addon.
  - The `metrics-server` is needed by e.g. the `horizontal-pod-autoscaler`.
  - The `dashboard` is a Web UI tool for managing a cluster (as an alternative to the CLI tool `kubectl`).
  - The `csi-hostpath-driver` is needed for `hostpath` storage when using multi-node Minikube clusters.
  - The `volumesnapshots` is needed by the `csi-hostpath-driver` for full functionality.
  - The `metallb` addon is a `load balancer` (layer 4 load balancer) for enabling external access (via the internet) to the Minikube cluster.
  - The `ingress` addon is an `nginx-ingress` (layer 7 load balancer) for enabling external access (via the internet) to the Minikube cluster.
- Addons can be disabled with the command `minikube addons disable <addon>`.

````python
#!minikube addons enable metrics-server -p minikube
!minikube addons enable metrics-server
!minikube addons enable dashboard
!minikube addons enable csi-hostpath-driver
!minikube addons enable volumesnapshots
!minikube addons enable metallb
!minikube addons enable ingress
````

## Stop the Minikube cluster

- The command `minikube stop -p <profile>` stops the Minikube cluster.
  - Where `<profile>` is the name of the Minikube profile, e.g. `minikube stop -p minikube`
  - Or just `minikube stop` for the default profile (which is called `minikube`).
  - It just stops the Docker containers for the Minikube nodes (since we are using Docker as the *driver*).
- A stopped (but not deleted) Minikube cluster, can be restarted with the command `minikube start -p <profile>`
  - Where `<profile>` is the name of the Minikube profile, e.g. `minikube start -p minikube`
  - Or just `minikube start` for the default profile.
  - This will restart the Minikube cluster with the same configuration it had when it was stopped.

````python
#!minikube stop -p minikube
!minikube stop
````

## Install the `metrics server`

The [Metrics Server](https://github.com/kubernetes-sigs/metrics-server) is a Kubernetes application used to scrape (collect) metrics from a Kubernetes cluster.

- We can use it to monitor server (node) and pod resource utilization (CPU, RAM, Storage, etc.).
- It's also a prerequisite for the `dashboard` and horizontal autoscaling (which we will cover in another workshop).

In the cells below, we use `Helm` (the Kubernetes package manager) to install the `metrics server`.

Let's start by listing `remote repositories` we have added to our `local repository`.
- Notice that the `local repository` is currently empty.

````python
!helm repo ls
````

Helm has an `Artifact Hub` that contains an index (information) about various Helm repositories (i.e. `remote repositories`).
- A `remote repository` contains various **Helm Charts**.
- A **Helm Chart** contains software we can install into a Kubernetes cluster.

Let's search for `metrics-server` on `Artifact Hub`.
  - We see that the `metrics-server` **Helm Chart** is available from numerous `remote repositories`.

````python
!helm search hub metrics-server --max-col-width 0
````

- Click on the first link under the `URL` column to visit the web page with information about that `remote repository`.
  
  https://artifacthub.io/packages/helm/metrics-server/metrics-server

- The web page tells use how to add the `remote repository` to our `local repository` (run the cell below to do this).
  - You can also visit the `remote repository's` web page for more information: https://kubernetes-sigs.github.io/metrics-server

````python
!helm repo add metrics-server https://kubernetes-sigs.github.io/metrics-server/
````

We can update our `local repositories` to their latest versions if their corresponding `remote repositories` have newer versions available.
- It's a good practice to do this after adding a `remote repository` to out `local repository` (although it might not always be necessary).

````python
!helm repo update
````

If we list our `local repositories` again, we see that the `metrics-server` repository has been added.

````python
!helm repo ls
````

We can search our `local repository` for **Helm Charts**.
- The `NAME` column shows the repository name and the chart name, i.e. `metrics-server/metrics-server`, where:
  - The first `metrics-server` before the `/` is the name of the `local repository`.
  - The second `metrics-server` after the `/` is the name of the `Helm Chart`.
- The `CHART VERSION` column shows the version of the Helm Chart (package).
- The `APP VERSION` column shows the version of the Kubernetes application (which is the actual `metrics-server` application).
- The `DESCRIPTION` column contains information about the Kubernetes application (i.e. the `metrics-server` application).

````python
# !helm search repo metrics-server
!helm search repo
````

We can also remove a `local repository`.
- Here `metrics-server` is the name of the repository.

````python
!helm repo rm metrics-server
````

Now, let's add the `metric-server` back into out `local repository`.

````python
!helm repo add metrics-server https://kubernetes-sigs.github.io/metrics-server/
!helm repo update
!helm repo ls
````

- Let's revisit the web page with information about the `remote repository` for the `metrics-server`.
  
  https://artifacthub.io/packages/helm/metrics-server/metrics-server

- The web page also tells use how to install the `metrics-server` **Helm Chart** into our Kubernetes cluster (run the cell below to do this).
  - I have added some additional flags:
    - `--set args="{--kubelet-insecure-tls}"` configures the `metrics-server` to **not** use TLS/HTTPS.
    - `--namespace kube-system` will install the `metrics-server` into the Kubernetes **namespace** `kube-system`.
    - `--create-namespace` will create the Kubernetes **namespace** if it doesn't already exist.

````python
!helm upgrade --install metrics-server metrics-server/metrics-server --set args="{--kubelet-insecure-tls}" --create-namespace --namespace kube-system
````

- We can list the **Helm Charts** we have installed into our Kubernetes cluster.
  - The `-A` flag lists **Helm Charts** in all Kubernetes **namespaces**.
  - Here we see that the `metrics-server` has been installed into the Kubernetes **namespace** `kube-system`.

````python
!helm list -A
````

- Let's verify the `metrics-server` was installed successfully.
  - The `kubectl top nodes` command returns metrics (cpu and memory utilization) for the Kubernetes nodes.
  - The `kubectl top pods -A` command returns metrics for the Pods in all namespaces (`-A`).
- Since the `top` command shows metrics information (CPU and Memory), the `metrics-server` was installed successfully.

````python
!kubectl top nodes
!kubectl top pods -A
````

We can also uninstall a **Helm Chart** from a Kubernetes cluster.
- Let's uninstall the `metrics-server` which is installed in the Kubernetes namespace `kube-system`.

````python
!helm uninstall metrics-server --namespace kube-system
````

Using a **YAML** file for configuring a **Helm Chart**.
- The command `helm show values <repository>/<chart>` returns configuration settings in YAML format for a **Helm Chart**, where:
  - `<repository>` is the name of the **local repository**, e.g. `metrics-server`
  - `<chart>` is the name of the **Helm Chart**, e.g. `metrics-server`

Let's create a folder `configuration` and *pipe* the returned YAML into a YAML file `metrics-server-values.yaml`.
- This will store the default configuration values for the `metrics-server` in the YAML file.

````python
!mkdir -p configuration
!helm show values metrics-server/metrics-server > configuration/metrics-server-values.yaml
````

We can supply a YAML configuration file when installing a **Helm Chart**.
- Open the file `configuration/metrics-server-values.yaml`.
- Change the line `args:[]` to:
  ```bash
  # args: []
  args:
  - --kubelet-insecure-tls
  ```
  - This is the *declarative* equivalent of the *imperative* setting `--set args="{--kubelet-insecure-tls}"` we previousy used when installing the `metrics-server`.
- Save the file, then use the `-f` flag to point to the YAML file `configuration/metrics-server-values.yaml` when installing the `metrics-server`.

````python
!helm upgrade --install metrics-server metrics-server/metrics-server --create-namespace --namespace kube-system -f configuration/metrics-server-values.yaml
````

Create a Service Account, Cluster Role Binding, and Token for the `dashboard`

- In order to access the `dashboard` we need to use Kubernetes' Role Based Access Control (RBAC) system.
  - Kubernetes RBAC is a topic which is out-of-scope for this course.
- We also need to generate a Bearer Token that we can use to login to the `dashboard`.

Execute the cell below, and copy the token that is returned on the last line in the output.

- We will need this token to login to the `dashboard`.

````python
!kubectl create serviceaccount kubeadmin -n kubernetes-dashboard
!kubectl create clusterrolebinding kubeadmin-binding --clusterrole=cluster-admin --serviceaccount=kubernetes-dashboard:kubeadmin
!kubectl -n kubernetes-dashboard create token kubeadmin
````

Open the Dashboard

- Run the following command in a separate terminal (and keep it open) to open a **proxy to the dashboard**:
  - `kubectl -n kubernetes-dashboard port-forward svc/kubernetes-dashboard-kong-proxy 8443:443`
- Visit: https://localhost:8443
- Enter the `Bearer Token` from above and click `Sign In` to login to the `dashboard`.

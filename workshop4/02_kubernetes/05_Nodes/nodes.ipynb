## List Nodes

- The command `kubectl get nodes` is used to list the nodes running in a kubernetes cluster.
  - Without the `-o wide` option, the command provides basic information about the nodes.
  - With the `-o wide` option, the command provides additional information about the nodes.
- Here we see:
  - The `NAME` of each node.
  - The `STATUS` of each node (here `Ready` means the node is ready to use).
  - The `ROLES` of each node (`control-plane` is a Control Plane node, whereas `worker` or `<none>` is a Worker node).
  - The `AGE` of each node, i.e. how long the node has been available.
  - The `VERSION` of kubernetes each node is using.
  - The `INTERNAL-IP` addess of each node (the internal IP address each node has within the kubernetes cluster).
  - The `EXTERNAL-IP` addess of each node (the public IP address each node has outside the kubernetes cluster).
  - The `OS-IMAGE` used by each node.
  - The `KERNEL-VERSION` of the OS used by each node.
  - The `CONTAINER-RUNTIME` used to run the containers.

- We have configured our Minikube Kubernetes cluster to use 3 nodes.
  - The node called `minikube` is a `control-plane` node (can also be used as a `worker` node in a Minikube cluster).
  - The nodes called `minikube-m02` and `minikube-m03` are `worker` nodes (`<none>` just means they haven't explicitly been labeled as `worker` nodes).

````python
#!kubectl get nodes
!kubectl get nodes -o wide
````

## Get information about a node

- The command `kubectl get node [NodeName]` returns information about a node with name `[NodeName]`.
  - The `-o yaml` option returns the information as a YAML manifest (definition) for a node.

- Some important information contained in a node listing:
  ```yaml
  apiVersion: v1
  kind: Node
  metadata:
    name: minikube            <--- the name of the control plane node will be "minikube" if you are using a default 3-node Minikube cluster
  status:
    addresses:
    - address: 192.168.49.2   <--- the node's IP address
      type: InternalIP
    - address: minikube       <--- the node's Hostname
      type: Hostname
    capacity:
      cpu: 8                           <--- number of CPU threads (number of CPU cores => divide by 2)
      ephemeral-storage: 101539832Ki   <--- storage (disk size) in kibibytes (in this case the total size of the physical disk since we are using a Docker driver)
      memory: 16354716Ki               <--- RAM in kibibytes (in this case the total size of the physical RAM since we are using a Docker driver)
      pods: 110                        <--- maximum number of Pods that the node can run
    images: <list>                     <--- list of container images downloaded to the node
    nodeInfo:
      architecture: amd64                                <--- container architecture (64-bit AMD system)
      containerRuntimeVersion: docker://29.2.1           <--- container runtime version
      kernelVersion: 6.17.0.14-generic                   <--- the OS kernel version (in this case a Linux kernel)
      kubeProxyVersion: ""                               <--- kube proxy version (in this case blank since we are using a Docker driver, but other might be e.g. v1.30.5)
      kubeletVersion: v1.35.1                            <--- kubelet version
      operatingSystem: linux                             <--- operating system type
      osImage: Debian GNU/Linux 12 (bookworm)            <--- the OS image used this node container is based on (in this case, Debian 12)
  ```

````python
!kubectl get node minikube -o yaml
# !kubectl get node minikube-m02 -o yaml
# !kubectl get node minikube-m03 -o yaml
````

## Describe a node

- The command `kubectl describe nodes` returns detailed information about all nodes.
- The command `kubectl describe node [NodeName]` returns detailed information about a node with name `[NodeName]`.
- A node description contains the same, but more detailed, information as a node's YAML listing above.
- In addition, a node description contains a list of important `Events` that have taken place (at the bottom of the listing).
 
  ```bash
  Events:
    Type    Reason                   Age                From             Message
    ----    ------                   ----               ----             -------
    Normal  RegisteredNode           17m                node-controller  Node minikube event: Registered Node minikube in Controller       
  ```

- The code below *describes* the control plane node.

````python
#!kubectl describe nodes
!kubectl describe node minikube
# !kubectl describe node minikube-m02
# !kubectl describe node minikube-m03
````

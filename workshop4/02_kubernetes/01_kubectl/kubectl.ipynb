## The `kubectl` CLI Tool

In this notebook, we look at the very basics of using the `kubectl` CLI tool, which is used to communicate with a Kubernetes cluster.

- Make sure you have a Minikube Kubernetes cluster running, e.g. `minikube start`.
- Also make sure you have installed the `kubectl` tool on your computer.

## Set `kubectl`'s context

Since `kubectl` can be used to communicate with multiple Kubernetes cluster's, we need to tell `kubectl` which cluster we want to communicate with.
- `kubectl config get-contexts` lists the available contexts.
- `kubectl config use-context minikube` switches to the `minikube` context.
- `kubectl config current-context` shows the currectly active context.

**Note**
- Minikube will automatically set the correct context when starting a Minikube cluster.
- Minikuve will automatically change back to the previous context when stopping a Minikube cluster.

```python
!kubectl config get-contexts
!kubectl config use-context minikube
!kubectl config current-context
```
